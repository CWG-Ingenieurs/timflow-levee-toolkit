Checks the geohydrological scheme (model parameters) of each transect, or creates an empty template to fill in.

| | |
|---|---|
| **Input** | `model_parameters_<transect>.xlsx`, tab `initial` (filled in by you) |
| **Output** | checked scheme per transect |

The scheme is in the tab `initial` of the parameter file: one row per zone (river, foreland, hinterland) and aquifer, with tops and bottoms, conductivity, resistance and storage. The tab `explanation` of the same file gives the meaning and unit of each column; it is shown in section 2. For a transect without a parameter file, this notebook creates an empty template (with both tabs) to fill in.

```python
from pathlib import Path

import pandas as pd

import timflow_levee_toolkit as tlt
tlt.show_versions()
```

```python
## UPDATE FOR YOUR PROJECT: the project folder (with project_settings.toml and the data)
project_dir = Path.cwd().parent / "examples" / "papendrecht"

project = tlt.load_project(project_dir)
units = project.units
project
```

## 1. Create templates for transects without a parameter file

```python
naq = 2  # number of aquifers in a new template

for transect in project.transects():
    path = project.parameters_file(transect)
    if not path.exists():
        tlt.create_parameter_template(path, naq=naq)
        print(f"{transect}: template created, fill in {project.relpath(path)}")
```

## 2. Columns of the scheme
The tab `explanation` of the parameter file describes each column of the tab `initial` (here from the file of the first transect).

```python
explanation = pd.read_excel(project.parameters_file(project.transects()[0]), sheet_name="explanation")
explanation.style.hide(axis="index")
```

## 3. Check the schemes

```python
for transect in project.transects():
    print(f"=== {transect}")
    try:
        inhom_params, naq = tlt.read_model_params(project.parameters_file(transect), tab_name="initial")
        tlt.validate_model_params(inhom_params, naq)
        tlt.print_model_params(inhom_params)
    except (ValueError, KeyError) as e:
        print(f"PROBLEM: {e}")
```
