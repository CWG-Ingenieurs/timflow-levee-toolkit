A project has one row per transect in the transects metadata. This notebook checks that table, creates the folders of all transects and shows which inputs are already available.

```python
from pathlib import Path

import timflow_levee_toolkit as tlt
```

## 1. Project
The project folder contains `project_settings.toml` with the folder and file names (optional; without it the defaults are used) and the input data.

```python
## UPDATE FOR YOUR PROJECT: the project folder (with project_settings.toml and the data)
project_dir = Path.cwd().parent / "examples" / "papendrecht"

project = tlt.load_project(project_dir)
units = project.units
project
```

## 2. Transects metadata
One row per transect. The sheet 'explanation' of the template describes the columns.

```python
metadata = project.metadata()

problems = tlt.validate_transects_metadata(metadata)
for problem in problems:
    print("PROBLEM:", problem)
if not problems:
    print("Transects metadata OK")

metadata
```

## 3. Create the folder structure

```python
project.create_folders()

for transect in project.transects():
    print(project.relpath(project.transect_dir(transect)))
```
